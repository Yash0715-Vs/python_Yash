## Q5 – Vehicle Inheritance

Create a parent class `Vehicle` and child class `Car`.

- `Vehicle` stores `name` and `speed`.
- `Car` adds `fuel_type`.
- Use `super()` in `Car`.
- `display()` should show all three values.

Fill only the missing logic.

```python
class Vehicle:
    def __init__(self, name, speed):
        self.name = name
        self.speed = speed

    def display(self):
        print(self.name, self.speed)


class Car(Vehicle):
    def __init__(self, name, speed, fuel_type):
        super().__init__(name,speed)
        self.fuel_type= fuel_type
        pass

    def display(self):
        print(self.name, self.speed, self.fuel_type)
        pass


car = Car("Creta", 180, "Petrol")
car.display()
```

## Q8 – Abstract Vehicle

Create an abstract `Vehicle` class.

Requirements:
- Use `ABC` and `@abstractmethod`.
- Create an abstract `start()` method.
- Create `Car` and `Bike` child classes.
- Each child must implement `start()`.
- Store both objects in a list and call `start()` using a loop.

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        print(F"THE VEHICAL IS STARTING")
        pass


class Car(Vehicle):
    def start(self):
        print(F"THE CAR IS STARTING")
        pass


class Bike(Vehicle):
    def start(self):
        print(F"THE BIKE IS STARTING")
        pass


vehicles = [Car(), Bike()]

for vehicle in vehicles:
    vehicle.start()
```

## Q9 – ATM System

Create an ATM system using inheritance, encapsulation, and polymorphism.

Requirements:
- Create a parent `Bank` class with private balance.
- Create `HDFC` and `AXIS` child classes.
- HDFC maximum withdrawal = ₹20,000.
- AXIS maximum withdrawal = ₹30,000.
- Withdrawal should happen only if both balance and bank limit are sufficient.
- Both child classes should override `withdraw()`.
- Use a loop to call `withdraw()`.

```python
class Bank:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance
        pass


class HDFC(Bank):
    def withdraw(self, amount):
        
        if amount <= 20000:
            print(f"withdrow successful {amount}")
        else:
            print("invalid")
        pass


class AXIS(Bank):
    def withdraw(self, amount):
        if amount <= 30000:
            print(f"withdrow successful {amount}")
        else:
            print("invalid")
        pass


banks = [
    HDFC(50000),
    AXIS(50000)
]

amounts = [15000, 25000]

for bank, amount in zip(banks, amounts):
    bank.withdraw(amount)
```

## Q10 – Food Delivery System

Create a food delivery system.

Requirements:
- Create an abstract class `Delivery`.
- Store `distance`.
- Create an abstract `delivery_time()` method.
- Create `Bike`, `Car`, and `Drone`.
- Bike: distance × 5 minutes.
- Car: distance × 3 minutes.
- Drone: distance × 1 minute.
- Store all objects in one list.
- Use one loop to display delivery time.

Try to complete the missing lines without changing the given structure.

```python
from abc import ABC, abstractmethod


class Delivery(ABC):
    def __init__(self, distance):
        self.distance = distance

    @abstractmethod
    def delivery_time(self):
        return 0
        pass


class Bike(Delivery):
    def delivery_time(self):
        return self.distance * 5
        pass


class Car(Delivery):
    def delivery_time(self):
        return self.distance * 3
        pass


class Drone(Delivery):
    def delivery_time(self):
        return self.distance * 1
        pass


deliveries = [
    Bike(4),
    Car(4),
    Drone(4)
]

for delivery in deliveries:
    print(delivery.delivery_time(), "minutes")
```
